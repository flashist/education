# Задача № 1

Компания решила провести A/B-тест новой рекламной кампании, предполагая, что новое объявление увеличит среднюю выручку на одного клиента с текущих 100 рублей до 110 рублей. Предполагая, что стандартное отклонение остается неизменным и составляет 20 рублей, определите размер выборки, необходимый для достижения мощности теста равной 80%, при уровне значимости α=0.05.

```python
import statsmodels.stats.power as smp

# Исходные данные
mean_control = 100   # Средняя выручка контрольной группы
mean_test = 110      # Средняя выручка тестовой группы
std_dev = 20         # Стандартное отклонение
power = 0.8          # Мощность теста
alpha = 0.05         # Уровень значимости

effect_size = abs(mean_control - mean_test) / std_dev  # Размер эффекта
sample_size = smp.TTestIndPower().solve_power(alpha=alpha, power=power, effect_size=effect_size)
print(f'Необходимый размер выборки: {round(sample_size)}')
```

#Задание №2

Задача:Проведено A/B-тестирование, направленное на повышение средней суммы покупки клиентов. Получены следующие данные:

- Контрольная группа (A): средняя сумма заказа составила 150 рублей, стандартное отклонение 25 рублей.
- Тестовая группа (B): средняя сумма заказа составила 160 рублей, стандартное отклонение 30 рублей.

Используя двусторонний t-тест, проверьте гипотезу о равенстве средних сумм заказов. Установите уровень значимости α=0.05, рассчитайте p-value и оцените значимость различий.

Задание:

```python
from scipy import stats
import numpy as np

# Данные групп
group_A_mean = 150
group_B_mean = 160
group_A_std = 25
group_B_std = 30
n_samples = 1000  # Количество наблюдений в каждой группе

# Генерируем случайные данные вокруг указанных средних и стандартных отклонений
data_group_A = np.random.normal(loc=group_A_mean, scale=group_A_std, size=n_samples)
data_group_B = np.random.normal(loc=group_B_mean, scale=group_B_std, size=n_samples)

# Двусторонний t-тест
t_statistic, p_value = stats.ttest_ind(data_group_A, data_group_B, equal_var=False)

print("T-статистика:", round(t_statistic, 4))
print("P-значение:", round(p_value, 4))
if p_value < 0.05:
    print("Различия значимы.")
else:
    print("Различия незначимы.")
```

#Задание №3

Задача:Для улучшения мобильного приложения разработано обновление, которое должно повысить среднее время активного взаимодействия пользователя с приложением. Компания произвела A/B-тестирование среди 500 пользователей, поделив их поровну на две группы. Были собраны следующие данные:

- Контрольная группа (A): среднее время активности 10 минут, стандартное отклонение 2 минуты.
- Тестовая группа (B): среднее время активности 11 минут, стандартное отклонение 2.5 минуты.

Определите, является ли разница в среднем времени активностью значимой на уровне α=0.05.

Задание:

```python
from scipy import stats

# Данные групп
group_A_mean = 10
group_B_mean = 11
group_A_std = 2
group_B_std = 2.5
n_samples = 250  # Количество наблюдений в каждой группе

# Генерация случайных данных
data_group_A = np.random.normal(loc=group_A_mean, scale=group_A_std, size=n_samples)
data_group_B = np.random.normal(loc=group_B_mean, scale=group_B_std, size=n_samples)

# Проведение t-теста
t_statistic, p_value = stats.ttest_ind(data_group_A, data_group_B, equal_var=False)

print("T-статистика:", round(t_statistic, 4))
print("P-значение:", round(p_value, 4))
if p_value < 0.05:
    print("Разница значима.")
else:
    print("Разница незначима.")
```

#Задание №4

Задача:Проверьте утверждение о том, что новый интерфейс веб-сайта влияет на коэффициент конверсии посетителей в покупателей. Текущие показатели:

- Контрольная группа (A): конверсия равна 10%.
- Тестовая группа (B): предполагаемая конверсия увеличилась до 12%.

Уровень значимости α=0.05, мощность теста β=0.8. Определите минимальный объем выборки для каждого сегмента.

Задание:

```python
import statsmodels.stats.api as sms

# Исходные данные
conversion_rate_A = 0.10
conversion_rate_B = 0.12
alpha = 0.05
power = 0.8

# Вычисляем эффект размера
effect_size = sms.proportion_effectsize(conversion_rate_B, conversion_rate_A)

# Расчет необходимого объема выборки
required_sample_size = sms.NormalIndPower().solve_power(power=power, alpha=alpha, effect_size=effect_size, alternative='two-sided')

print(f"Требуемое количество наблюдений на группу: {round(required_sample_size)}")
```

#Задание №5

Задача:Проанализируем гипотезу о влиянии нового функционала мобильной игры на увеличение количества игровых сессий игроков. Проводится A/B-тест среди 1000 пользователей. Собраны такие данные:

- Контрольная группа (A): среднее количество сессий — 5, стандартное отклонение — 1.
- Тестовая группа (B): среднее количество сессий — 5.5, стандартное отклонение — 1.2.

Проверяем гипотезу на уровне значимости α=0.05.

Задание:

```python
from scipy import stats

# Данные групп
group_A_mean = 5
group_B_mean = 5.5
group_A_std = 1
group_B_std = 1.2
n_samples = 1000  # Количество наблюдений в каждой группе

# Генерация случайных данных
data_group_A = np.random.normal(loc=group_A_mean, scale=group_A_std, size=n_samples)
data_group_B = np.random.normal(loc=group_B_mean, scale=group_B_std, size=n_samples)

# Проведение t-теста
t_statistic, p_value = stats.ttest_ind(data_group_A, data_group_B, equal_var=False)

print("T-статистика:", round(t_statistic, 3))
print("P-значение:", round(p_value, 3))
if p_value < 0.05:
    print("Гипотеза отвергается, различия значимы.")
else:
    print("Гипотеза принимается, различия незначимы.")
```
