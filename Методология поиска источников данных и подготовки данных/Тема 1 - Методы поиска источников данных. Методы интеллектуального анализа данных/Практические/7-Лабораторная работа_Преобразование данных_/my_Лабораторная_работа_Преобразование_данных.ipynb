# 1. Преобразовать таблицу с использованием .pivot_table

### Задание:
Используя библиотеку `pandas`, выполните следующие шаги для анализа данных о продажах товаров в различных регионах России:

1. **Создание набора данных**: создайте словарь с информацией о продажах, включающий регионы (`'region'`), категории продуктов (`'product_category'`), месяцы продажи (`'month'`), выручку (`'revenue'`), количество проданных единиц товара (`'quantity'`) и себестоимость продукции (`'cost_price'`).
   
2. **Преобразование данных в DataFrame**: преобразуйте созданный словарь в объект `DataFrame`.

3. **Расчет дополнительных показателей**:
   - Рассчитайте валовую прибыль (`'margin'`) путем вычета себестоимости из выручки.
   - Определите среднюю цену единицы товара (`'average_price'`) делением общей выручки на количество проданного товара.

4. **Формирование сложной сводной таблицы**: используя метод `.pivot_table()`, постройте отчет, где строки будут содержать комбинацию региона и категории продукта, столбцы — месяца продаж, а значения будут включать суммы выручки, прибыли и среднее значение средней цены.

5. **Агрегационные функции**: применяйте соответствующие агрегирующие функции:
   - Для выручки и прибыли — суммирование значений.
   - Для средней цены — вычисление среднего арифметического значения с округлением до двух знаков после запятой.

6. **Дополнительные настройки отчета**: добавьте строку общих итогов и заполните пустые ячейки нулями.

7. **Форматирование числовых значений**: примените специальную функцию форматирования чисел, чтобы отображались знаки разделения тысяч и фиксированное число десятичных разрядов.

8. **Стилизация отчета**: установите заголовок для сводной таблицы.

Результатом задания должна стать красивая и наглядная таблица отчёта по продажам с возможностью визуального восприятия всех ключевых метрик.


Задание:

````python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

# Набор данных
sales_data = {
    'region': ['Москва', 'Санкт-Петербург', 'Новосибирск', 'Екатеринбург']*3,
    'product_category': ['Электроника', 'Одежда', 'Продукты']*4,
    'month': ['январь', 'февраль', 'март']*4,
    'revenue': [10000, 5000, 3000, 8000, 6000, 4000]*2,
    'quantity': [100, 50, 150, 120, 80, 100]*2,
    'cost_price': [5000, 3000, 2000, 4000, 3500, 2500]*2
}

df_sales = pd.DataFrame(sales_data)

# Дополнительные расчёты
df_sales['margin'] = df_sales['revenue'] - df_sales['cost_price']
df_sales['average_price'] = df_sales['revenue'] / df_sales['quantity']

# Сложная сводная таблица
# Для average_price используем lambda с округлением до 2 знаков
pivoted_df = df_sales.pivot_table(
    index=['region', 'product_category'],
    columns='month',
    values=['revenue', 'margin', 'average_price'],
    aggfunc={'revenue': 'sum', 'margin': 'sum', 'average_price': lambda x: round(x.mean(), 2)},
    margins=True,
    fill_value=0
)

month_order = list(dict.fromkeys(df_sales['month'])) + ['All']
pivoted_df = pivoted_df.reindex(columns=pivoted_df.columns.reindex(month_order, level=1)[0])

# Форматирование по колонкам
def custom_formatter(val):
  if isinstance(val, (float, int)):
      return f"{val:,.2f}".replace(",", " ")
  else:
      return val

# Приложение формата ко всей таблице
formatted_pivoted_df = pivoted_df.applymap(custom_formatter)

# После форматирования применяем стилизатор
styled_pivoted_df = formatted_pivoted_df.style.set_caption("Отчёт по продажам")

# Показываем итоговую таблицу
styled_pivoted_df
````

# 2. Переформатировать данные с помощью .melt

## Задача:
Создать аналитический отчёт по зарплате сотрудников компании за первые полгода с использованием библиотеки Pandas в Python.

### Описание задачи:
У вас имеется исходный набор данных о сотрудниках, содержащий следующую информацию:
- Идентификатор сотрудника (`employee_id`)
- Должность сотрудника (`position`)
- Зарплата сотрудника за каждый месяц первого полугодия (`Jan_salary`, `Feb_salary`, ..., `Jun_salary`)

Вам необходимо произвести ряд операций над этими данными для последующего формирования подробного отчёта по зарплатам сотрудников.

### Последовательность действий:

#### Этап 1: Создание и подготовка данных
- Создайте фрейм данных (DataFrame), соответствующий предоставленной структуре.
- Данные должны отражать информацию о нескольких сотрудниках разных должностей и их заработных платах за шесть первых месяцев года.

#### Этап 2: Преобразование структуры данных («расплавка»)
- Используйте метод `.melt()` для преобразования широкого формата данных в узкий (long format). Это позволит вам удобно анализировать динамику изменения зарплат по каждому сотруднику.
- Сделайте так, чтобы столбец с названием месяца назывался `'month'`, а столбец с значениями заработной платы — `'salary'`.  

#### Этап 3: Изменение названий месяцев
- Приведите название каждого месяца к русскоязычному формату («январь», «февраль» и т.п.). Используйте для этого словарь соответствий между оригинальными именами столбцов и переведёнными наименованиями.

#### Этап 4: Расчёт квартальной зарплаты
- Подсчитайте суммарную зарплату каждого сотрудника за первый квартал (январь–март) и сохраните её в отдельном столбце под именем `'Q1_salary'`.

#### Этап 5: Объединение результатов
- Добавьте рассчитанные квартальные итоги в общий DataFrame путём объединения данных методом `.merge()`.

#### Этап 6: Возвращение широкой формы представления данных
- Превратите полученные данные снова в широкий формат (wide format) с помощью метода `.pivot_table()`, формируя таблицу таким образом, чтобы строками были сотрудники и должности, а столбцами — месяцы.

#### Итоговая задача:
Полученный отчёт должен демонстрировать зарплаты сотрудников за каждый месяц и сумму первой четверти отдельно. Постарайтесь оформить вывод аккуратно и понятно, сделав акцент на удобстве чтения итогового результата.

Ваш конечный продукт — сформированный отчёт по зарплатам сотрудников за первое полугодие, включая квартальную статистику, представленную в удобной форме.

Задание:

````python
import pandas as pd

# Генерируем больший и более разнообразный набор данных
salary_data = {
    'employee_id': [1, 2, 3, 4, 5],
    'position': ['Менеджер', 'Разработчик', 'Дизайнер', 'HR', 'Маркетолог'],
    'Jan_salary': [50000, 60000, 45000, 40000, 55000],
    'Feb_salary': [55000, 65000, 50000, 45000, 60000],
    'Mar_salary': [52000, 62000, 48000, 42000, 58000],
    'Apr_salary': [58000, 68000, 52000, 47000, 62000],
    'May_salary': [60000, 70000, 55000, 50000, 65000],
    'Jun_salary': [62000, 72000, 58000, 52000, 68000]
}

df_salary = pd.DataFrame(salary_data)

# Шаг 1: Расплавим данные (из широких в длинные)
melted_df = df_salary.melt(
    id_vars=['employee_id', 'position'],  # Оставляем employee_id и position неизменными
    var_name='month',                      # Название нового столбца с месяцем
    value_name='salary'                   # Название нового столбца с зарплатой
)

# Шаг 2: Преобразуем названия месяцев для удобства (из Jan_salary → январь)
months_map = {
    'Jan_salary': 'январь',
    'Feb_salary': 'февраль',
    'Mar_salary': 'март',
    'Apr_salary': 'апрель',
    'May_salary': 'май',
    'Jun_salary': 'июнь'
}
melted_df['month'] = melted_df['month'].map(months_map)

# Шаг 3: Посчитаем квартальную зарплату (за первый квартал января-марта)
q1_months = ['январь', 'февраль', 'март']
quarterly_salary = (
    melted_df[melted_df['month'].isin(q1_months)]
    .groupby('employee_id', as_index=False)['salary']
    .sum()
    .rename(columns={'salary': 'Q1_salary'})
)

# Шаг 4: Добавим квартальную зарплату обратно в основной DataFrame
final_df = melted_df.merge(quarterly_salary, on='employee_id', how='left')

# Шаг 5:

# Выводим основную таблицу с помесячными зарплатами
final_wide_df = final_df.pivot_table(
    index=['employee_id', 'position'],
    columns='month',
    values='salary',
    aggfunc='sum',
    fill_value=0
)

month_order = list(months_map.values())  # порядок из уже существующего словаря
final_wide_df = final_wide_df[month_order]

final_wide_df.index.names = ['ID сотрудника', 'Должность']
final_wide_df.columns.name = None

print("Зарплаты по месяцам:")
print(final_wide_df)

# Отдельная таблица с итогами за первый квартал
q1_summary = final_df.groupby(['employee_id', 'position'])['Q1_salary'].first().reset_index()
q1_summary.columns = ['ID сотрудника', 'Должность', 'Зарплата за Q1 (январь–март)']

print("\nИтоги за первый квартал:")
print(q1_summary.to_string(index=False))
````

## Задача:
Необходимо проанализировать заказы интернет-магазина и составить полный отчёт по ним, объединив информацию о самих заказах и покупателе. Важно учесть случаи, когда некоторые клиенты отсутствуют в списке покупателей или информация о заказе неполная (например, дата доставки неизвестна).

### Исходные данные:
Имеются два набора данных:
1. **Набор данных о заказах (`orders`)**, который включает такие поля:
   - Уникальный номер заказа (`order_id`);
   - ID покупателя (`customer_id`), который может отсутствовать, если заказ сделан анонимно или отменён;
   - Дата оформления заказа (`order_date`);
   - Дата доставки (`delivery_date`), которая также может быть пустой, если доставка ещё не осуществлена;
   - Сумма заказа (`amount`);
   - Статус платежа (`payment_status`: оплачено, отменено, ожидает оплаты);
   - Размер скидки (`discount`).

2. **Набор данных о покупателях (`customers`)**, содержащий такую информацию:
   - ID покупателя (`customer_id`);
   - Имя покупателя (`name`);
   - E-mail адрес покупателя (`email`);
   - Уровень лояльности покупателя (`loyalty_level`: золото, серебро, бронза).

### Требования к выполнению:
1. Провести объединение данных о заказах и покупателях, учитывая возможные пропуски в обоих наборах данных. Если покупатель указан, но заказ отсутствует или наоборот, эта информация тоже должна учитываться.

2. Проверить наличие клиентов в обеих таблицах и отметить недействительные заказы (если статус оплаты заказа равен `"canceled"` или клиент не найден в базе).

3. Составить полную статистическую картину по каждому клиенту:
   - Количество сделанных заказов;
   - Общую сумму покупок;
   - Общий размер скидок, полученных данным клиентом;
   - Число действительно выполненных заказов (оплаченных и подтвержденных).

4. Подготовить понятный и читаемый отчёт по всем заказам, объединённым с клиентами, а также итоговые статистики по каждому покупателю.

### Что должно присутствовать в отчёте?
1. Полный список заказов вместе с дополнительной информацией о каждом клиенте (имя, e-mail, уровень лояльности).
2. Отдельная статистика по каждому клиенту, показывающая ключевые показатели (количество заказов, общая сумма расходов, общие скидки, число действительных заказов).

Таким образом, ваш финальный отчёт должен представлять собой детальное представление о текущих заказах магазина и характеристиках поведения покупателей.


Задание:

````python
import pandas as pd

# Больший объём данных и дополнительные атрибуты
orders = {
    'order_id': [1, 2, 3, 4, 5],
    'customer_id': [101, 102, 101, 103, None],  # customer_id может отсутствовать
    'order_date': ['2023-01-01', '2023-01-02', '2023-01-03', '2023-01-04', '2023-01-05'],
    'delivery_date': ['2023-01-05', '2023-01-06', '2023-01-07', '2023-01-08', None],  # Возможны пропущенные delivery_date
    'amount': [1000, 1500, 2000, 1200, 1800],
    'payment_status': ['paid', 'canceled', 'pending', 'paid', 'paid'],  # Новый атрибут статуса оплаты
    'discount': [100, 0, 50, 0, 0]  # Атрибут скидки
}

customers = {
    'customer_id': [101, 102, 103],
    'name': ['Иван Иванов', 'Анна Петрова', 'Сергей Смирнов'],
    'email': ['ivan@example.com', 'anna@example.com', 'sergey@example.com'],
    'loyalty_level': ['gold', 'silver', 'bronze']  # Уровень лояльности покупателя
}

# Загружаем данные в датафреймы
df_orders = pd.DataFrame(orders)
df_customers = pd.DataFrame(customers)

# Решим проблему слияния, учитывая возможность отсутствия customer_id
# Либо у заказчика отсутствует информация, либо заказ отменён
merged_df = pd.merge(df_orders, df_customers, on='customer_id', how='outer', indicator=True)

merged_df['customer_id'] = merged_df['customer_id'].astype('Int64')

# Добавим дополнительный столбец для проверки действительных заказов
# Действительный заказ: статус 'paid' и клиент найден в базе
merged_df['valid_order'] = (merged_df['payment_status'] != 'canceled') & (merged_df['_merge'] == 'both')

# Подчистим лишние колонки (_merge)
merged_df.drop('_merge', axis=1, inplace=True)

# Сгруппируем данные по клиентам и посмотрим общую статистику по каждому покупателю
summary_stats = merged_df.groupby('customer_id').agg({
    'order_id': 'count',
    'amount': 'sum',
    'discount': 'sum',
    'valid_order': 'sum'
}).rename(columns={
    'order_id': 'total_orders',
    'amount': 'total_spent',
    'discount': 'total_discount',
    'valid_order': 'valid_orders'
})

# Сделаем удобный вывод данных
print("\nПолный отчёт по заказам:")
print(merged_df.to_string())

print("\nСтатистика по покупателям:")
print(summary_stats)
````

````python
import pandas as pd

# Составим более богатый набор данных (продажи за несколько месяцев, магазины и категории товаров)
sales_data = {
    'store': ['Магазин 1', 'Магазин 1', 'Магазин 1', 'Магазин 2', 'Магазин 2', 'Магазин 2'],
    'month': ['Январь', 'Февраль', 'Март', 'Январь', 'Февраль', 'Март'],
    'category': ['Фрукты', 'Овощи', 'Напитки', 'Фрукты', 'Овощи', 'Напитки'],
    'total_revenue': [10000, 6000, 4000, 8000, 5000, 3000],
    'units_sold': [1000, 500, 400, 800, 500, 300]
}

df_sales = pd.DataFrame(sales_data)

col_names = {
    'store': 'Магазин',
    'category': 'Категория',
    'month': 'Месяц',
    'profit_per_unit': 'Прибыль на ед.',
    'growth_rate': 'Прирост прибыли (%)'
}

# Шаг 1: Создадим колонку profit_per_unit (прибыль на единицу товара)
df_sales['profit_per_unit'] = (df_sales['total_revenue'] / df_sales['units_sold']).round(2)

print("Шаг 1 — прибыль на единицу товара:")
print(df_sales[['store', 'category', 'month', 'profit_per_unit']].rename(columns=col_names))

# Шаг 2: Проследим динамику прироста прибыли относительно предыдущего периода
df_sales['previous_profit'] = df_sales.groupby(['store', 'category'])['profit_per_unit'].shift(1)
df_sales['growth_rate'] = (
    (df_sales['profit_per_unit'] - df_sales['previous_profit'])
    / df_sales['previous_profit'] * 100
).round(1)

nan_count = df_sales['growth_rate'].isna().sum()
print(f"\nШаг 2 — динамика прироста прибыли:")
print(df_sales[['store', 'category', 'month', 'profit_per_unit', 'growth_rate']].rename(columns=col_names))

print("\n")
if nan_count == len(df_sales):
    print("⚠ growth_rate не удалось рассчитать: для каждой комбинации магазин+категория присутствует только одна запись.")
elif nan_count > 0:
    print(f"⚠ growth_rate не удалось рассчитать для {nan_count} строк: отсутствуют данные за предыдущий период.")

# Шаг 3: Проведем многоуровневую группировку с named aggregation
grouped_results = df_sales.groupby(['store', 'category']).agg(
    avg_profit_per_unit=('profit_per_unit', 'mean'),
    avg_growth_rate=('growth_rate', 'mean'),
    total_units=('units_sold', 'sum'),
    total_revenue=('total_revenue', 'sum')
).round(2)

# Шаг 4: Выведем итоговую сводную таблицу
print("\nСводная итоговая таблица")
print(grouped_results.rename(columns={
    'avg_profit_per_unit': 'Ср. прибыль на ед.',
    'avg_growth_rate': 'Ср. прирост (%)',
    'total_units': 'Всего единиц',
    'total_revenue': 'Выручка'
}).rename_axis(index={'store': 'Магазин', 'category': 'Категория'}))
````
