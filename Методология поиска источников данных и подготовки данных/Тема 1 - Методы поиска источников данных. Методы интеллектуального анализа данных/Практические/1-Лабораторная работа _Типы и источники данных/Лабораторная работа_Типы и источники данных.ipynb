```python
import csv
from datetime import datetime

def filter_customers_by_purchase_date(csv_file_path, date):
    with open(csv_file_path, 'r', newline='') as file:
        reader = csv.DictReader(file)
        filtered_customers = []

        for row in reader:
            purchase_date = datetime.strptime(row['purchase_date'], '%Y-%m-%d')

            if purchase_date > datetime.strptime(date, '%Y-%m-%d'):
                filtered_customers.append(row['name'])

    return len(filtered_customers), filtered_customers

# Тестируем с синтетическим набором данных
data = '''id,name,email,purchase_date
1,Alexey,a@a.ru,2023-02-15
2,Olga,o@o.ru,2023-01-10
3,Pavel,p@p.ru,2023-03-05'''

with open('customers.csv', mode='w') as f:
    f.write(data)

count, customers = filter_customers_by_purchase_date('customers.csv', '2023-01-31')
print(f'Количество записей: {count}')
print('Покупатели:', ', '.join(customers))
```

```python
import csv

def find_duplicate_emails(csv_file_path):
    emails_count = {}

    with open(csv_file_path, 'r', newline='') as file:
        reader = csv.DictReader(file)

        for row in reader:
            email = row['email']
            emails_count[email] = emails_count.get(email, 0) + 1

    duplicates = [(email, count) for email, count in emails_count.items() if count > 1]

    if duplicates:
        print("Обнаружены следующие дублирующиеся электронные адреса:")
        for dup_email, count in duplicates:
            print(f'{dup_email}, количество повторений: {count}')
    else:
        print("Дублирующих электронных адресов не найдено.")

# Тестируем с синтетическим набором данных
data = '''participant_id,name,email
1,Alexey,a@a.ru
2,Olga,b@b.ru
3,Pavel,c@c.ru
4,Dmitry,d@d.ru
5,Sergei,a@a.ru'''  # Добавляем повторяющийся e-mail

with open('participants.csv', mode='w') as f:
    f.write(data)

find_duplicate_emails('participants.csv')
```

Решение:

```python
import csv

def average_order_value(csv_file_path):
    total_orders = 0
    total_sum = 0
    registered_total = 0
    guest_total = 0
    registered_count = 0
    guest_count = 0

    with open(csv_file_path, 'r', newline='') as file:
        reader = csv.DictReader(file)

        for row in reader:
            order_sum = float(row['order_sum'])
            is_registered = int(row['is_registered'])  # 1 — зарегистрированный клиент, 0 — гость

            total_orders += 1
            total_sum += order_sum

            if is_registered:
                registered_total += order_sum
                registered_count += 1
            else:
                guest_total += order_sum
                guest_count += 1

    avg_total = round(total_sum / total_orders, 2)
    avg_registered = round(registered_total / registered_count, 2) if registered_count != 0 else None
    avg_guests = round(guest_total / guest_count, 2) if guest_count != 0 else None

    print(f"Средняя сумма заказа всего: {avg_total}")
    print(f"Средняя сумма заказа для зарегистрированных пользователей: {avg_registered or 'Нет данных'}")
    print(f"Средняя сумма заказа для гостей: {avg_guests or 'Нет данных'}")

# Тестируем с синтетическим набором данных
data = '''order_id,user_id,is_registered,order_sum
1,1,1,5000
2,2,0,3000
3,3,1,7000
4,4,0,2000'''

with open('orders_data.csv', mode='w') as f:
    f.write(data)

average_order_value('orders_data.csv')
```
