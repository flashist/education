## 1. Простое чтение и фильтрация данных из CSV-файла

Цель: Считайте файл формата CSV, содержащий информацию о клиентах интернет-магазина («id», «имя», «электронная почта», «дата последней покупки»). Затем выведите количество записей и имена всех пользователей, чьи последние покупки были сделаны позднее определённой даты (например, после 2023-01-31).

Задание:

```python
import csv
from datetime import datetime

def filter_customers_by_purchase_date(csv_file_path, date):
    cutoff = datetime.strptime(date, "%Y-%m-%d").date()
    filtered_customers = []

    with open(csv_file_path, mode="r", newline="", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        for row in reader:
            purchase_date = datetime.strptime(row["purchase_date"], "%Y-%m-%d").date()
            if purchase_date > cutoff:
                filtered_customers.append(row["name"])

    return len(filtered_customers), filtered_customers

# Тестируем с синтетическим набором данных
data = '''id,name,email,purchase_date
1,Alexey,a@a.ru,2023-02-15
2,Olga,o@o.ru,2023-01-10
3,Pavel,p@p.ru,2023-03-05'''

with open('customers.csv', mode='w') as f:
    f.write(data)

count, customers = filter_customers_by_purchase_date('customers.csv', '2023-01-31')
print(f'Количество записей: {count}')
print('Покупатели:', ', '.join(customers))
```

## 2. Подсчет количества покупок по категориям

Цель: Вы загружаете CSV-файл, содержащий столбцы «id», «название продукта», «категория», «стоимость». Ваша задача — посчитать общее количество покупок каждой категории и вывести топ-3 категорий по числу покупок.

Задание:

```python
import csv

def count_purchases_by_category(csv_file_path):
    category_counts = {}

    with open(csv_file_path, mode="r", newline="", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        for row in reader:
            category = row["category"].strip()
            category_counts[category] = category_counts.get(category, 0) + 1

    top3 = sorted(category_counts.items(), key=lambda x: (-x[1], x[0]))[:3]

    print("Топ-3 категорий по числу покупок:")
    for category, count in top3:
        print(f"{category}: {count}")

# Тестируем с синтетическим набором данных
data = '''id,title,category,price
1,Laptop,Electronics,80000
2,Mobile Phone,Electronics,30000
3,T-shirt,Clothing,1500
4,Jacket,Clothing,5000
5,Book,School Supplies,700
6,Notebook,School Supplies,1000'''

with open('products.csv', mode='w') as f:
    f.write(data)

count_purchases_by_category('products.csv')
```

## 3. Агрегация данных по регионам

Цель: Есть CSV-файл с колонками «имя покупателя», «регион проживания», «сумма заказа». Требуется рассчитать суммарную выручку для каждого региона и выбрать регион с наибольшими продажами.

Задание:

```python
import csv

def calculate_revenue_per_region(csv_file_path):
  region_sum = {}

  # Ваш код здесь...
  with open(csv_file_path, mode="r", newline="", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
      region = row["region"].strip()
      order_amount = float(row["order_amount"].strip())
      region_sum[region] = region_sum.get(region, 0) + order_amount

  top1 = sorted(region_sum.items(), key=lambda x: (-x[1], x[0]))[:1][0]

  print("Топ-1 регион по выручке:")
  print(f"{top1[0]}: {top1[1]}")

# Тестируем с правильными заголовками и синтетическим набором данных
data = '''customer_name,region,order_amount
Alexey,Moscow,10000
Olga,St.Petersburg,8000
Pavel,Krasnodar,5000
Dmitry,Novosibirsk,12000
Anna,Yekaterinburg,9000'''

with open('orders.csv', mode='w') as f:
    f.write(data)

calculate_revenue_per_region('orders.csv')
```

## 4. Поиск повторяющихся e-mail адресов

Цель: У вас имеется CSV-файл с данными о посетителях конференции, включая уникальные идентификационные номера участников и адреса электронной почты. Задача состоит в выявлении дублирующихся e-mail адресов среди участников и формировании отчёта по ним.


Задание:

```python
import csv

def find_duplicate_emails(csv_file_path):
    emails_count = {}

    with open(csv_file_path, mode="r", newline="", encoding="utf-8") as f:
      reader = csv.DictReader(f)
      for row in reader:
        email = row["email"].strip()
        emails_count[email] = emails_count.get(email, 0) + 1

    duplicates_exist = any(count > 1 for count in emails_count.values())

    if duplicates_exist:
      print("Дублирующиеся email:")
      for email, count in emails_count.items():
        if count > 1:
          print(f"{email}: {count}")
    else:
      print("Дублирующих электронных адресов не найдено.")

# Тестируем с синтетическим набором данных
data = '''participant_id,name,email
1,Alexey,a@a.ru
2,Olga,b@b.ru
3,Pavel,c@c.ru
4,Dmitry,d@d.ru
5,Sergei,a@a.ru'''  # Добавляем повторяющийся e-mail

with open('participants.csv', mode='w') as f:
    f.write(data)

find_duplicate_emails('participants.csv')
```

## 5. Расчёт средней суммы заказа

Цель: Вам предоставлен CSV-файл с заказами интернет-магазина («номер заказа», «идентификатор покупателя», «сумма заказа», «дата оформления»). Рассчитайте среднюю сумму одного заказа отдельно для зарегистрированных пользователей и гостей (неавторизованных пользователей).

```python
import csv

def average_order_value(csv_file_path):
    total_orders = 0
    total_sum = 0
    registered_orders = 0
    registered_sum = 0
    guests_total = 0
    guests_sum = 0

    # Ваш код здесь...

    with open(csv_file_path, mode="r", newline="", encoding="utf-8") as f:
      reader = csv.DictReader(f)
      for row in reader:
        user_id = row["user_id"].strip()
        is_registered = row["is_registered"] == "1"
        order_sum = float(row["order_sum"])

        total_orders += 1
        total_sum += order_sum
        if is_registered:
          registered_orders += 1
          registered_sum += order_sum
        else:
          guests_total += 1
          guests_sum += order_sum

    avg_total = round(total_sum / total_orders, 2) if total_orders > 0 else total_sum
    avg_registered = round(registered_sum / registered_orders, 2) if registered_orders > 0 else registered_sum
    avg_guests = round(guests_sum / guests_total, 2) if guests_total > 0 else guests_sum

    print(f"Средняя сумма заказа всего: {avg_total or 'Нет данных'}")
    print(f"Средняя сумма заказа для зарегистрированных пользователей: {avg_registered or 'Нет данных'}")
    print(f"Средняя сумма заказа для гостей: {avg_guests or 'Нет данных'}")

# Тестируем с синтетическим набором данных
data = '''order_id,user_id,is_registered,order_sum
1,1,1,5000
2,2,0,3000
3,3,1,7000
4,4,0,2000'''

with open('orders_data.csv', mode='w') as f:
    f.write(data)

average_order_value('orders_data.csv')
```
