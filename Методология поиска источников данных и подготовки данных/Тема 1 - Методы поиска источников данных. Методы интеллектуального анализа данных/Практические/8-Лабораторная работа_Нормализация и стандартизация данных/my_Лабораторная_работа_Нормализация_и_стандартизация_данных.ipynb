# 1: Min-Max Нормализация

## **Задание**

Напишите программу, реализующую процесс нормализации входных данных методом **Min-Max**. Затем выполните сравнение результатов ручной реализации алгоритма с функцией из библиотеки `scikit-learn`. Программа должна выводить оба результата нормализации.

### Входные данные:
Набор данных представлен следующим двумерным массивом NumPy:

$$
\begin{bmatrix}
-1 & 2 \\
-0.5 & 6 \\
0 & 10 \\
1 & 18
\end{bmatrix}
$$

### Требования к программе:
1. Напишите функцию `manual_min_max`, которая принимает массив данных и возвращает матрицу с элементами, приведенными к диапазону $[0, 1]$ по каждому столбцу отдельно.
   
   Формула для расчета Min-Max нормализации по одному столбцу выглядит так:
   
$$
x_{norm} = \frac{x - x_{\text{min}}}{x_{\text{max}} - x_{\text{min}}}
$$

   Где:
   - $x$ — значение элемента,
   - $x_{\text{min}}$ — минимальное значение в данном столбце,
   - $x_{\text{max}}$ — максимальное значение в данном столбце.

2. Выполните самостоятельную реализацию, а также проведите аналогичную операцию с использованием встроенного класса `MinMaxScaler` из библиотеки `scikit-learn`.

3. Выведите обе нормализованные матрицы на экран и убедитесь, что ваши результаты совпадают с результатами стандартной реализации.

### Ожидаемый вывод программы:
Программа должна вывести две таблицы нормализованных значений: одну, рассчитанную вами вручную, и вторую, полученную с помощью библиотеки `scikit-learn`.

Пример вывода:

```
Ручная нормализация:
[[0.         0.        ]
 [0.33333333 0.28571429]
 [0.66666667 0.57142857]
 [1.         1.        ]]

MinMaxScaler (scikit-learn):
[[0.         0.        ]
 [0.33333333 0.28571429]
 [0.66666667 0.57142857]
 [1.         1.        ]]
```

Убедитесь, что ваша реализация совпадает с результатом работы встроенной функции.


Задание:

````python
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# Исходные данные
data = np.array([[-1, 2], [-0.5, 6], [0, 10], [1, 18]])

def manual_min_max(data):
    # Нормализация вручную через явный цикл по столбцам
    col_min = data.min(axis=0)
    col_max = data.max(axis=0)
    normalized = np.zeros_like(data, dtype=float)
    for j in range(data.shape[1]):
        for i in range(data.shape[0]):
            normalized[i, j] = (data[i, j] - col_min[j]) / (col_max[j] - col_min[j])
    return normalized

# Ручная нормализация
manual_result = manual_min_max(data)
print('Ручная нормализация:\n', manual_result)

# Проверка с помощью scikit-learn
scaler = MinMaxScaler()
sklearn_result = scaler.fit_transform(data)
print('\nMinMaxScaler (scikit-learn):\n', sklearn_result)

print('\nОбе матрицы идентичны, что говорит о одинаковом поведении встроенной функции и нашей реализации')
````

# 2: Z-нормализация (стандартизация)

## Задание

Реализуйте алгоритм Z-нормализации (или стандартного отклонения) вручную и сравните полученный результат с реализацией из библиотеки Scikit-Learn (`StandardScaler`).

### Условия:
Вам дан следующий набор данных:

$$
\begin{bmatrix}
-1 & 2 \\
-0.5 & 6 \\
0 & 10 \\
1 & 18
\end{bmatrix}
$$

Вашей задачей является выполнение следующей последовательности действий:

1. Самостоятельно напишите функцию `manual_z_normalization()`, которая реализует формулу Z-нормализации по следующему принципу:

$$
x' = \frac{(x - \mu)}{\sigma},
$$

где:
- $x'$ — нормализованное значение,
- $\mu$ — среднее арифметическое элементов соответствующего столбца,
- $\sigma$ — стандартное отклонение элементов соответствующего столбца.

Формулы вычисления среднего ($\mu$) и стандартного отклонения ($\sigma$) по столбцам:

$$
\mu_j = \frac{\sum_{i=1}^{N}x_{ij}}{N},\quad\sigma_j=\sqrt{\frac{\sum_{i=1}^{N}(x_{ij}-\mu_j)^2}{N}},
$$

где $j$ обозначает номер столбца, $i$ — строку, $N$ — количество строк.

2. После самостоятельной реализации примените библиотеку Scikit-Learn (`StandardScaler`) для сравнения результатов.

3. Отобразите результаты обеих нормализаций и убедитесь, что они совпадают.

### Требуемый вывод:
Вы получите две нормализованные матрицы: одна, созданная вручную, вторая — с помощью `StandardScaler`. Приведите результаты вашего кода и сделайте вывод о совпадении результатов двух подходов.

Пример ожидаемого вывода:

```
Z-нормализация ручная:
[[-1.22474487 -1.34164079]
 [-0.40824829 -0.4472136 ]
 [ 0.40824829  0.4472136 ]
 [ 1.22474487  1.34164079]]

StandardScaler (scikit-learn):
[[-1.22474487 -1.34164079]
 [-0.40824829 -0.4472136 ]
 [ 0.40824829  0.4472136 ]
 [ 1.22474487  1.34164079]]
```

Таким образом, ваше решение должно подтвердить правильность реализации Z-нормализации и показать согласованность результатов.

Задание:

````python
from sklearn.preprocessing import StandardScaler

# Данные остаются теми же
data = np.array([[-1, 2], [-0.5, 6], [0, 10], [1, 18]])

def manual_z_normalization(data):
    n = data.shape[0]
    # Среднее: сумма / N по каждому столбцу
    col_mean = data.sum(axis=0) / n
    # Стандартное отклонение по формуле из задания (ddof=0)
    col_std = np.sqrt(((data - col_mean) ** 2).sum(axis=0) / n)
    return (data - col_mean) / col_std

# Ручная нормализация
z_norm_manual = manual_z_normalization(data)
print('Z-нормализация ручная:\n', z_norm_manual)

# Использование scikit-learn
scaler = StandardScaler()
z_norm_sklearn = scaler.fit_transform(data)
print('\nStandardScaler (scikit-learn):\n', z_norm_sklearn)

# Вывод о совпадении результатов
if np.allclose(z_norm_manual, z_norm_sklearn):
  print("\nВывод: результаты совпадают — ручная реализация корректна.")
else:
  print("\nВывод: результаты расходятся — в реализации есть ошибка.")
````

````python
from sklearn.preprocessing import RobustScaler

# Создаем новый массив с несколькими аномально большими значениями ("выбросы")
outlier_data = np.array([[1, 2], [2, 4], [3, 6], [100, 200]])  # Последние два значения — выбросы

print('Исходные данные:\n', outlier_data)

# Робастная нормализация
robust_scaler = RobustScaler()
robust_scaled_data = robust_scaler.fit_transform(outlier_data)
print('\nРобастная нормализация (RobustScaler):\n', robust_scaled_data)

# Сравнение с другими методами
mm_scaled = MinMaxScaler().fit_transform(outlier_data)
std_scaled = StandardScaler().fit_transform(outlier_data)
print('\nMin-Max (для сравнения):\n', mm_scaled)
print('\nStandardScaler (для сравнения):\n', std_scaled)

print('\n--- Анализ ---')
print('RobustScaler выбран потому, что данные содержат выбросы.')
print('Метод использует медиану и межквартильный размах (IQR = Q3 - Q1),')
print('которые нечувствительны к экстремальным значениям.')
print()
print('Сравнение: выброс сильно искажает Min-Max (все нормальные')
print('значения сжимаются к 0) и StandardScaler (смещает среднее и std).')
print('RobustScaler масштабирует нормальные строки в разумном диапазоне,')
print('а выброс просто получает большое значение, не искажая остальные.')
````

   - Используем встроенный датасет "Iris" из библиотеки Scikit-Learn.

   - Реализуйте три вида нормализации с помощью соответствующих классов из модуля `sklearn.preprocessing`: `MinMaxScaler`, `StandardScaler`, `RobustScaler`.

   - Посмотрите на первые пять строк каждой нормализованной таблицы.

````python
import pandas as pd
from sklearn.datasets import load_iris

# Загрузка датасета Iris
X, _ = load_iris(return_X_y=True)
df = pd.DataFrame(X, columns=['sepal_length', 'sepal_width', 'petal_length', 'petal_width'])

# Применение трёх методов нормализации
scaled_mm= MinMaxScaler().fit_transform(df)
scaled_std = StandardScaler().fit_transform(df)
scaled_rb= RobustScaler().fit_transform(df)

print('Min-Max Normalized Data (первые 5 строк):')
print(scaled_mm[:5])

print('\nZ-Normalized Data (первые 5 строк):')
print(scaled_std[:5])

print('\nRobust Scaled Data (первые 5 строк):')
print(scaled_rb[:5])

# Проверка правильности применения каждого метода
print("\n--- Проверка корректности трансформации ---")

# Min-Max: все значения должны быть в диапазоне [0, 1]
print(f"Min-Max: min={scaled_mm.min():.4f}, max={scaled_mm.max():.4f}"
  f"(ожидается: min=0.0, max=1.0)")

# StandardScaler: среднее ≈ 0, std ≈ 1 по каждому столбцу
print(f"Z-Scale: mean={scaled_std.mean(axis=0).round(4)}, "
  f"std={scaled_std.std(axis=0).round(4)}"
  f"(ожидается: mean≈0, std≈1)")

# RobustScaler: медиана ≈ 0 по каждому столбцу
print(f"Robust:median={np.median(scaled_rb, axis=0).round(4)}"
  f"(ожидается: median≈0)")

print('\n--- Анализ ---')
print('Min-Max приводит все значения к диапазону [0, 1], но чувствителен к выбросам:')
print('один экстремальный признак сожмёт остальные значения к нулю.')
print()
print('StandardScaler центрирует данные (среднее=0, std=1), что хорошо для большинства')
print('алгоритмов, но среднее и std смещаются при наличии выбросов.')
print()
print('RobustScaler использует медиану и IQR, поэтому устойчив к выбросам.')
print('Для датасета Iris, где признаки имеют разные масштабы и возможны выбросы,')
print('RobustScaler является наиболее надёжным выбором.')
````

````python
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier

# Загружаем датасет Iris и разделяем на признаки (X) и целевую переменную (y)
X, y = load_iris(return_X_y=True)

# Классификационные модели
classifiers = {
    'Logistic Regression': LogisticRegression(max_iter=10000),
    'SVM':                 SVC(kernel='linear'),
    'Random Forest':       RandomForestClassifier(n_estimators=100, random_state=42),
}

# Методы нормализации данных
scalers = {
    'Min-Max': MinMaxScaler(),
    'Z-Scale': StandardScaler(),
    'Robust':  RobustScaler(),
}

# Результаты собираем в список и выводим таблицей
results = []
for clf_name, classifier in classifiers.items():
    for scale_name, scaler in scalers.items():
        scaled_data = scaler.fit_transform(X)
        # Кросс-валидация с 5 фолдами
        scores = cross_val_score(classifier, scaled_data, y, cv=5)
        results.append({
            'Классификатор':    clf_name,
            'Нормализация':     scale_name,
            'Средняя точность': round(scores.mean(), 3),
        })

results_df = pd.DataFrame(results)
print(results_df.to_string(index=False))

# Интерпретация результатов
best_score = results_df['Средняя точность'].max()
best_rows = results_df[results_df['Средняя точность'] == best_score]

print(f"\n--- Интерпретация результатов ---")
for clf_name in results_df['Классификатор'].unique():
  clf_df = results_df[results_df['Классификатор'] == clf_name]
  best_score = clf_df['Средняя точность'].max()
  best_scalers = clf_df[clf_df['Средняя точность'] == best_score]['Нормализация'].tolist()
  print(f"{clf_name}: лучшая нормализация — {', '.join(best_scalers)} "
        f"(точность: {best_score})")

print()
print("Факторы, влияющие на различия результатов:")
print("- SVM и LogisticRegression чувствительны к масштабу признаков, поэтому")
print("  нормализация существенно влияет на их точность.")
print("- Random Forest основан на деревьях решений и инвариантен к масштабу,")
print("  поэтому результаты слабо зависят от метода нормализации.")
print()
print("Рекомендации:")
print("- Для линейных моделей (LR, SVM) предпочтительны Z-Scale или Min-Max.")
print("- При наличии выбросов в данных лучше использовать RobustScaler.")
print("- Random Forest можно применять без нормализации, результат будет схожим.")
````

- Min-Max: приводит значения к диапазону [0, 1]
- Z-Scale (StandardScaler): центрирует данные (среднее=0, std=1)
- Robust (RobustScaler): использует медиану и IQR, устойчив к выбросам

Рассматриваются три классификатора:
- Logistic Regression — линейная модель
- SVM с линейным ядром — линейная модель, чувствительная к масштабу
- Random Forest — ансамблевая модель на основе деревьев решений

Оценка производится методом кросс-валидации с 5 фолдами.

### Результаты эксперимента
- Logistic Regression показала лучший результат с Z-Scale (0.960)
- SVM показала лучший результат с Min-Max и Z-Scale (0.967)
- Random Forest показала одинаковый результат со всеми методами (0.967)

### Анализ и выводы
Logistic Regression наиболее чувствительна к выбору нормализации: разброс
результатов составил 0.033 (от 0.927 до 0.960). Z-Scale оказался оптимальным,
так как центрирование данных улучшает сходимость линейных моделей.

SVM также чувствителен к масштабу, однако показал высокую точность с Min-Max
и Z-Scale. Robust дал чуть меньший результат (0.960).

Random Forest инвариантен к масштабу признаков — все три метода нормализации
дали одинаковый результат (0.967), что подтверждает теоретические свойства
алгоритма на основе деревьев решений.

Рекомендация: для линейных моделей предпочтителен Z-Scale или Min-Max.
При наличии выбросов в данных следует использовать RobustScaler.
Для Random Forest выбор метода нормализации не имеет значения.
